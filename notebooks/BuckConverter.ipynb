```python
from ml4cps import examples, vis, famos, tools
```

```python
_, data = examples.buck_converter()
change_points = [d[0] for d in data]
```

```python
data = [d[2] for d in data]
```

```python
fig = vis.plot_timeseries(data, modedata=change_points, showlegend=True)
fig.update_layout(height=1000)
```

```python
# Normalize data
data = famos.normalize(data)
```

```python
fig = vis.plot_timeseries(data, modedata=change_points, showlegend=True)
fig.update_layout(height=1000)
```

```python
data = tools.extend_derivative(data, use_derivatives=[0, 1])
```

```python
WINDOW_SIZE = 10
change_points = []
for d in data:
    chp = famos.detect_change_points(d, window_size=WINDOW_SIZE)
    change_points.append(chp)
    
```
