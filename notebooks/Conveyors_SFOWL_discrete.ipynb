```python
from ml4cps import examples
from ml4cps.automata import learn
from ml4cps import tools
from ml4cps import vis
```

```python
discrete_data, cont_data = examples.conveyor_system_sfowl()
discrete_data_changes = tools.remove_timestamps_without_change(discrete_data, sig_names=discrete_data[0].columns)
discrete_data_changes[0].head()
```

## Visualizing simple automaton

```python
ta = learn.simple_learn_from_signal_vectors(discrete_data_changes)
ta.remove_rare_transitions(min_num=10)
fig = vis.plot_cps_plotly(ta, show_num_occur=True, show_events=True, show_state_label=False)
fig.update_layout(height=1000)

fig.show()
```

```python
fig = vis.plot_cps_component(ta, output='notebook', freq_as_edge_thickness=True, show_transition_freq=True)
```
